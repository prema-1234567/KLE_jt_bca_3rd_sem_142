```python
def student(*marks):
    sum = 0
    avg=len(marks)

    for i in marks:
        sum = sum + i
    print("sum =", sum)
    print("avg value", sum/avg)
    print("the max value is", max(marks))
    print("the min value is", min(marks))
student(1, 2, 3, 4, 5)
student(5,10,25,40,55,70,95,110)
```
